Getting the best models on the planet to generate an SVG

Inspired by the legendary Simon Willison's Pelican riding a bike test

Key point: this is a very different task to image generation! The model needs to describe the image with lines and shapes.

### This uses OpenRouter.ai so that we easily access the latest models

```python
from dotenv import load_dotenv
from IPython.display import Markdown, display
from datetime import datetime
import time
from revealer import reveal
from openai import OpenAI
import os
load_dotenv()
```

```python
# OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1/openai"

OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")
if OPENROUTER_API_KEY and OPENROUTER_API_KEY.startswith("sk-or-"):
    print("OPENROUTER_API_KEY looks good so far")
else:
    print("OPENROUTER_API_KEY doesn't seem right")
```

```python
openrouter = OpenAI(base_url=OPENROUTER_BASE_URL, api_key=OPENROUTER_API_KEY)
```

```python
challenge = "a panda rollerblading to work"
prompt = f"Generate an SVG of {challenge}. Respond with the SVG only, no code blocks."
messages = [{"role": "user", "content": prompt}]
```

```python
def artist(model, effort=None):
    try:
        start = datetime.now()
        
        # Prepare arguments: only add effort if it's an OpenAI reasoning model
        kwargs = {
            "model": model,
            "messages": messages,
        }
        if effort and ("o1" in model or "o3" in model or "gpt-5" in model):
            kwargs["reasoning_effort"] = effort

        response = openrouter.chat.completions.create(**kwargs)
        result = response.choices[0].message.content
        
        end = datetime.now()
        elapsed = (end - start).total_seconds()
        heading = f"### {model}\n**Time:** {elapsed // 60:.0f} min {elapsed % 60:.0f} s\n\n"
        return heading, result

    except Exception as e:
        # This catches the HTML 404 error and prints a cleaner message
        error_msg = str(e)[:100] # Limit length in case it's a huge HTML block
        print(f"Model {model} failed: {error_msg}")
        return f"### {model}\n**Error:** {error_msg}\n\n", None
```

```python
import os
from datetime import datetime
from openai import OpenAI

# 1. Setup Client
OPENROUTER_BASE_URL = "https://openrouter.ai/api/v1"
OPENROUTER_API_KEY = os.getenv("OPENROUTER_API_KEY")

client = OpenAI(base_url=OPENROUTER_BASE_URL, api_key=OPENROUTER_API_KEY)

def artist(model):
    try:
        start = datetime.now()
        
        # Note: We use images.generate for actual image creation
        response = client.images.generate(
            model=model,
            prompt="a panda rollerblading to work, cinematic lighting, high resolution",
            n=1,
            size="1024x1024"
        )
        
        # OpenRouter returns the URL of the generated image
        image_url = response.data[0].url
        
        end = datetime.now()
        elapsed = (end - start).total_seconds()
        
        heading = f"### {model}\n**Time:** {elapsed // 60:.0f} min {elapsed % 60:.0f} s\n"
        return heading, image_url

    except Exception as e:
        print(f"Model {model} failed. Error: {str(e)[:50]}")
        return f"### {model}\n**Error:** Model unavailable or invalid slug.\n", None

# 2. Correct Model Slugs for Image Generation (Current for 2026)
# These are the actual IDs used by OpenRouter for image models
# Use these slugs if your account balance is $0.00
results = [
    artist("google/gemini-2.0-flash-exp:free"),
    artist("deepseek/deepseek-chat:free"),
    artist("meta-llama/llama-3.3-70b-instruct:free"),
    artist("mistralai/mistral-7b-instruct:free")
]

# 3. Print the Results
for heading, url in results:
    print(heading)
    if url:
        print(f"Image Link: {url}\n")
```
